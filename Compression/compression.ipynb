```python
# Globals

# Imports
import heapq
from collections import defaultdict
```

```python
class MinHeapNode:
    def __init__(self, data, freq):
        self.left = None
        self.right = None
        self.data = data
        self.freq = freq

    def __lt__(self, other):
        return self.freq < other.freq

def encode_huffman(text):
    if not text:
        return "", None
    freq = defaultdict(int)
    for char in text:
        freq[char] += 1
    if len(freq) == 1:
        char = list(freq.keys())[0]
        root = MinHeapNode(char, freq[char])
        return "0" * len(text), root
    minHeap = []
    for char, frequency in freq.items():
        minHeap.append(MinHeapNode(char, frequency))
    heapq.heapify(minHeap)
    
    while len(minHeap) > 1:
        left = heapq.heappop(minHeap)
        right = heapq.heappop(minHeap)
        merged = MinHeapNode('$', left.freq + right.freq)
        merged.left = left
        merged.right = right
        heapq.heappush(minHeap, merged)
    
    root = minHeap[0]
    
    codes = {}

    def generateHuffmanCodes(node, code):
        if node is None:
            return
        if node.data != '$':  # Leaf node
            codes[node.data] = code if code else "0"
        generateHuffmanCodes(node.left, code + "0")
        generateHuffmanCodes(node.right, code + "1")
    generateHuffmanCodes(root, "")
    encoded_string = ""
    for char in text:
        encoded_string += codes[char]
    return encoded_string, root

def decode_huffman(encoded_string, huffman_tree_root):
    if not encoded_string or huffman_tree_root is None:
        return ""
    if huffman_tree_root.left is None and huffman_tree_root.right is None:
        return huffman_tree_root.data * len(encoded_string)
    decoded_string = ""
    current = huffman_tree_root
    for bit in encoded_string:
        if bit == '0':
            current = current.left
        else:
            current = current.right
        if current.left is None and current.right is None:
            decoded_string += current.data
            current = huffman_tree_root
    
    return decoded_string
```

```python
test_string = "Geek for Greeks"    
encoded, tree_root = encode_huffman(test_string)
print(f"Encoded: {encoded}")
print(len(encoded) / (len(test_string)))
    
# Decode
decoded = decode_huffman(encoded, tree_root)
print(f"Decoded: '{decoded}'")
print(f"Encoding/Decoding successful: {test_string == decoded}")
freq = defaultdict(int)
for char in test_string:
    freq[char] += 1
'''
codes = {}
def get_codes(node, code):
    if node is None:
        return
    if node.data != '$':
        codes[node.data] = code if code else "0"
    get_codes(node.left, code + "0")
    get_codes(node.right, code + "1")
    
get_codes(tree_root, "")
    
print("\nCharacter frequencies and codes:")
for char in sorted(codes.keys()):
    print(f"'{char}': frequency={freq[char]}, code='{codes[char]}'")'''
```
